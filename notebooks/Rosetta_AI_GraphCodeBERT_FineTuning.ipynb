```python
# Step 3: Run GraphCodeBERT Fine-Tuning Pipeline
import os
from pathlib import Path
from src.embeddings.codebert.finetune import train_contrastive_graphcodebert

results = train_contrastive_graphcodebert(
    data_file=Path('data/curated/parallel_corpus.jsonl'),
    epochs=3,
    batch_size=8,
    learning_rate=2e-5
)

print('=' * 60)
print('ROSETTA AI GRAPHCODEBERT FINE-TUNING RESULTS')
print('=' * 60)
print('BEFORE Fine-Tuning Equivalent Pair Cosine Sim:', results['before']['avg_equivalent_sim'])
print('BEFORE Fine-Tuning Random Pair Cosine Sim:    ', results['before']['avg_random_sim'])
print('AFTER Fine-Tuning Equivalent Pair Cosine Sim: ', results['after']['avg_equivalent_sim'])
print('AFTER Fine-Tuning Random Pair Cosine Sim:     ', results['after']['avg_random_sim'])
print('Delta Cosine Similarity Improvement:          ', results['after']['delta'] - results['before']['delta'])
print('=' * 60)
```
